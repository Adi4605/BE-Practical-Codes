```python
# Simple DNA Sequence Analysis (works with a FASTA file) 
# Put example_dna.fa in the same folder as this script. 
  
import re 
import matplotlib.pyplot as plt 
  
# ---------- 1. Read DNA sequence from FASTA file ---------- 
sequence = "" 
with open("example_dna.fa") as f: 
   for line in f: 
       if not line.startswith(">"):          # skip the header line 
           sequence += line.strip().upper() 
  
# ---------- 2. Basic information ---------- 
length = len(sequence) 
A = sequence.count("A") 
T = sequence.count("T") 
G = sequence.count("G") 
C = sequence.count("C") 
  
print("DNA Sequence:", sequence) 
print("Length:", length) 
print("A:", A, "T:", T, "G:", G, "C:", C) 
  
# ---------- 3. GC content ---------- 
gc = (G + C) / length * 100 
print("GC Content:", round(gc, 2), "%") 
  
# ---------- 4. Motifs (positions are 1-based) ---------- 
print("\nMotifs:") 
for motif in ["ATG", "TATA", "CG"]: 
   positions = [m.start() + 1 for m in re.finditer(f"(?={motif})", sequence)] 
   print(motif, ": count =", len(positions), ", positions =", positions) 
  
# ---------- 5. Coding regions (ATG ... stop codon) ---------- 
print("\nCoding Regions (at least 30 codons):") 
found = 0 
for i in range(len(sequence) - 2): 
   if sequence[i:i+3] == "ATG": 
       for j in range(i + 3, len(sequence) - 2, 3): 
           if sequence[j:j+3] in ["TAA", "TAG", "TGA"]: 
               orf = sequence[i:j+3] 
               if len(orf) >= 90:          
                   print(f"Position {i+1}-{j+3} ({len(orf)//3} codons):", orf) 
                   found += 1 
               break 
if found == 0: 
   print("No coding region found.") 
  
# ---------- 6. Plot ---------- 
bases = ["A", "T", "G", "C"] 
counts = [A, T, G, C] 
  
plt.bar(bases, counts, color=["green", "red", "orange", "blue"]) 
plt.title("DNA Base Composition") 
plt.xlabel("Base") 
plt.ylabel("Count") 
plt.show() 
 
 
```
