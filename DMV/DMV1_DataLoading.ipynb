```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1. Load (CSV given; make Excel and JSON copies to practice all 3 formats)
csv_df = pd.read_csv("sales.csv")
csv_df.to_excel("sales.xlsx", index=False)
csv_df.to_json("sales.json", orient="records")
excel_df = pd.read_excel("sales.xlsx")
json_df = pd.read_json("sales.json")
```

```python
# 2. Explore
print(csv_df.shape)
print(csv_df.head())
csv_df.info()
print(csv_df.isnull().sum())
print(csv_df.duplicated().sum())
```

```python
# 3. Clean
df = csv_df.drop_duplicates()
df = df.dropna()
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
# 4. Unify (all 3 formats in one dataframe)
all_df = pd.concat([csv_df, excel_df, json_df], ignore_index=True).drop_duplicates()
print(all_df.shape)
```

```python
# 5. Transform
df["Year"] = df["Order Date"].dt.year
df["Delivery Days"] = (df["Ship Date"] - df["Order Date"]).dt.days
df["Profit Margin %"] = df["Total Profit"] / df["Total Revenue"] * 100
```
