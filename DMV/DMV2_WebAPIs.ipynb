```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
API_KEY = "YOUR_API_KEY"
cities = ["Pune", "Mumbai", "Delhi", "Chennai", "Kolkata"]
```

```python
# 2. Call the API for each city
rows = []
for city in cities:
    url = "https://api.openweathermap.org/data/2.5/forecast"
    params = {"q": city, "appid": API_KEY, "units": "metric"}
    data = requests.get(url, params=params).json()
    lat, lon = data["city"]["coord"]["lat"], data["city"]["coord"]["lon"]
    
    # 3. Extract attributes
    for item in data["list"]:
        rows.append({
            "city": city, "lat": lat, "lon": lon,
            "time": item["dt_txt"],
            "temp": item["main"]["temp"],
            "humidity": item["main"]["humidity"],
            "wind": item["wind"]["speed"],
            "rain": item.get("rain", {}).get("3h", 0),
        })
df = pd.DataFrame(rows)
```

```python
# 4. Clean
df["time"] = pd.to_datetime(df["time"])
df = df.drop_duplicates().fillna(0)
df["date"] = df["time"].dt.date
```

```python
# 5. Model / statistics
print(df[["temp", "humidity", "wind", "rain"]].describe())
print(df.groupby("city")["temp"].agg(["mean", "max", "min"]))
```

```python
# 7. Aggregate daily
daily = df.groupby(["city", "date"])[["temp", "humidity", "wind", "rain"]].mean().reset_index()
print(daily.head())
```

```python
# 6. Visualize (Pune)
pune = df[df["city"] == "Pune"]
plt.plot(pune["time"], pune["temp"]); plt.title("Pune Temperature"); plt.xticks(rotation=45); plt.show()
daily.groupby("date")["rain"].sum().plot(kind="bar", title="Daily Rainfall"); plt.show()
plt.scatter(pune["temp"], pune["humidity"]); plt.xlabel("Temp"); plt.ylabel("Humidity"); plt.show()
```

```python
# 8. Geographical plot
geo = df.groupby(["city", "lat", "lon"])["temp"].mean().reset_index()
plt.scatter(geo["lon"], geo["lat"], c=geo["temp"], s=300, cmap="coolwarm")
plt.colorbar(label="Avg Temp (C)")
for _, r in geo.iterrows():
    plt.text(r["lon"], r["lat"], r["city"])
plt.show()
```

```python
# 9. Correlation heatmap
sns.heatmap(df[["temp", "humidity", "wind", "rain"]].corr(), annot=True, cmap="coolwarm")
plt.show()
```
