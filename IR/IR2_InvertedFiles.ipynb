```python
import re
from collections import defaultdict
```

```python
# Collection of docs
documents = {
    1: "Python is a popular programming language",
    2: "Python is used for machine learning and data science",
    3: "Machine Learning is a part of artificial intelligence",
    4: "Data Science uses Python for data analysis"
}
```

```python
# Create Inverted Index
inverted_index = defaultdict(set)

for doc_id, text in documents.items():
    #Convert to lower
    text = text.lower()
    # Extract words
    words = re.findall(r'\b[a-z]+\b', text)
    # Add document ID for each word
    for word in words:
        inverted_index[word].add(doc_id)
```

```python
# Display inverted index
print("Inverted Index")
print("--------------")
for word in sorted(inverted_index):
    print(word, '->', sorted(inverted_index[word]))
```

```python
# Search for a query
query = input("\nEnter search query: ").lower()
query_words = re.findall(r'\b[a-z]+\b', query)
# find docs containing query words
result_docs = set()
for word in query_words:
    if word in inverted_index:
        result_docs.update(inverted_index[word])


# Display Retrieved Docs
print("\nRetrieved Documents")
print("---------------------")

if result_docs:
    for doc_id in sorted(result_docs):
        print("Document ", doc_id,": ", documents[doc_id])
else:
    print("No matching documents found")
```
